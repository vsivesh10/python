```python
#sets in python
#Set is a collection of unique values that is unordered and mutable
numbers={10, 20, 30, 20, 10}
print(numbers)
```

```python
#why use set?
#suppose students have selected subjects
subjects = {"Python", "Java", "Python"}
print(subjects)
```

```python
#add values to a set
subjects = {"Python", "SQL"}
subjects.add("JAVA")
print(subjects)
```

```python
#remove values from a set
subjects.remove("JAVA")
print(subjects)
```

```python
#sets do not allow duplicate values
numbers={1, 2, 2, 3, 3, 4}
print(numbers)
```
