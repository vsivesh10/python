```python
#list in python
#list is an ordered and changeable collection that can store multiple values
marks = [80, 90, 75, 85]
print(marks)
```

```python
#accessing elements in a list
marks = [80, 90, 75, 85]
print(marks[0])
print(marks[1])
print(marks[3])
```

```python
#change elemts in a list
marks = [80, 90, 75]
marks[1] = 95
print(marks)
```

```python
#change elements in a list
marks = [75,95,86,70,69,98]
marks[5] = 200
marks[3] = 100
marks[1] = 90
print(marks)
```

```python
#accessing the elements
print(marks[0])
print(marks[5])
```

```python
#add elements to a list
marks = [75,95,86,70,69,98]
marks.append(100)
print(marks)
```

```python
#remove elements from a list
marks = [80, 90, 75]
marks.remove(90)
print(marks)
```

```python
#insert in a list
numbers = [10, 20, 30]
numbers.insert(1, 15)
print(numbers)
```

```python
#and bitwise operator
a = 7
b = 3
print(a & b)
```

```python
#or operator bitwise
a = (6)
b = (2)
print(a | b)
```

```python
#extend
a = [1, 2, 3]
b = [4, 5, 6]
a.extend(b)
print(a)
```

```python
#clear list
numbers = [1,2,3,4,5]
numbers.clear()
print(numbers)
```

```python
#index value of data
numbers = [10, 20, 30, 40]
print(numbers.index(30))
```

```python
#counting numbers
numbers = [10, 20, 20, 30]
print(numbers.count(20))
```

```python
#increasing order
numbers = [30, 50, 10, 20]
numbers.sort()
print(numbers)
```

```python
#decreasing order
numbers = [10, 20, 30, 40, 50]
numbers.sort(reverse=True)
print(numbers)
```

```python
#reverse
numbers = [10, 20, 30, 40]
numbers.reverse()
print(numbers)
```

```python
#copy 
a = [1, 2, 3]
b = a.copy()
print(b)
```

```python
numbers = [10, 20, 30, 40, 50]
print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
print(numbers[::-1])
```
