```python
#tuples in python
#Tuple is a collection of multiple values that is ordered and cannot be changed after creation
student = ("SIVESH", 20, "Python")
print(student[0])
```

```python
#immutable nature of tuples
student = ("SIVESH", 20, "Python")
student[1] = 22
#this gives an error because we cannot modify a tuple
```

```python
#tuples are immutable, meaning they cannot be changed after they are created. they are defined
numbers = (10,20,20,30,20)
print(numbers.count(20))
```

```python
numbers = (10,20,30,40)
print(numbers.index(30))
```

```python
numbers = (10, 20, 30, 40)
print(len(numbers))
print(max(numbers))
print(min(numbers))
print(sum(numbers))
```
