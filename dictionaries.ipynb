```python
#dictionaries in python
#dictionary is a collection of key-value pairs that is unordered and mutable
student = {
    "name" : "Sivesh",
    "age" : 20,
    "course" : "Python"
}
print(student)
```

```python
#access elements in dictionary
print(student["name"])
print(student["age"])
print(student["course"])
```

```python
#add new data to a dictionary
student["city"] = "Vijayawada"
print(student)
```

```python
#remove data
student.pop("city")
print(student)
```

```python
print(student.keys())
#keys() returns all the keys in dictionary
print(student.values())
#values() returns all the values in the dictionary
print(student.items())
#items() returns all key-value pairs
print(student.get("name"))
#get() returns the value of the specified key
student.update({"age":22})
#update() updates the value of the specified key
print(student)
student.pop("age")
#pop removes the specified key and its value
print(student)
```

```python
#popitem() removes the last inserted key value pair
student = {
    "name" : "Sivesh",
    "age" : 20,
    "course" : "Python"
}
student.popitem()
print(student)
```

```python
student = {
    "name" : "Sivesh"
}
student.setdefault("age",21)
print(student)
```
