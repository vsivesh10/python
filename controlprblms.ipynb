```python
#find the largest number among 5 numbers entered by the user
largest = None
for i in range(5):
    number=int(input("Enter number:"))
    if largest is None or number>largest:
        largest=number
print("Largest:",largest)
```

```python
#count the number of positive, negative and zero numbers entered by the use
positive = 0
negative = 0
zero = 0
for i in range(10):
    number = int(input("Enter number:"))
    if number>0:
        positive = positive + 1
    elif number<0:
        negative = negative + 1
    else:
        zero = zero + 1
print("Positive:",positive)
print("Negative:",negative)
print("Zero:",zero)
```

```python
#calculate the sum of positive number entered by the user, ignoring negative numbers
total = 0
for i in range(10):

    number = int(input("Enter a number:"))
    if number<0:
        continue
    total = total+number
print("Total",total)
```
