```python
#function for even/odd
def check_even_odd(number):
    if number%2==0:
        return "Even"
    else:
        return "odd"
result = check_even_odd(25)
print(result)
```

```python
#functions for largest of two numbers
def largest(a, b):
    if a>b:
        return a
    else:
        return b
result = largest(50, 30)
print("Largest:",result)
```

```python
#function + user input
def square(number):
    return number*number
number = int(input("Enter a number:"))
result = square(number)
print("Square:",result)
```

```python
#function+loop
def multiplication_table(number):
    for i in range(1,11):
        print(number,"x",i,"=",number*i)
number = int(input("Enter number:"))
multiplication_table(number)
```

```python
#add two numbers
def add(a,b):
    return a+b
x = int(input("Enter first number:"))
y = int(input("Enter second number"))
result = add(x,y)
print("Sum=",result)
```

```python
#largest of three numbers
def largest(a,b,c):
    if a>=b and a>=c:
        return a
    elif b>=a and b>=c:
        return b
    else:
        return c
a = int(input("Enter a: "))
b = int(input("Enter b: "))
c = int(input("Enter c: "))
print("Largest=", largest(a, b, c))
```

```python
def find_grade(marks):
    if marks>=90:
        return "A"
    elif marks>=80:
        return "B"
    elif marks>=70:
        return "C"
    elif marks>=60:
        return "D"
    else:
        return "Fail"
marks=int(input("Enter marks:"))
print("Grade=",find_grade(marks))
```
